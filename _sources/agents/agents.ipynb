# NGSolve for people and agents

## How we use Claude on the Netgen/NGSolve code base

Netgen/NGSolve is C++ with Python bindings, grown over more than 20 years. Agents now work on it with us:

* **Read and explain legacy code**: what an old routine does, who calls it, which invariants it relies on
* **Design notes first**: the agent writes a plan, a person reviews it, and only then code changes
* **Rework with verification**: one topic per commit, each change A/B-tested against the old code (for the mesher: bit-identical meshes), the agent prepares the change and the commit message, a person reviews and commits
* **Take over contributions**: e.g. the speed-ups of the `faster_netgen` fork, measured 2–3$\times$ faster mesh generation (M4 Pro, 16 test cases), re-implemented step by step in our code base
* **Technical notes**: a series of short notes on methods and features of NGSolve, written for people *and* agents: [docu.ngsolve.org technotes](https://github.com/NGSolve/ngsolve/tree/master/docs/technotes)

## Example: a Halloween ghost (Technical Note 2026-04)

<video src="ghost.mp4" width="300" controls autoplay loop muted></video>

A sheet with eyes and mouth falls onto an invisible sphere and is dragged around by it. Done on October 31, together with an agent, from the idea to the movie.

### What is inside

* nonlinear shell: hybridized **Hellan–Herrmann–Johnson** bending, membrane strain by **Regge interpolation** (no membrane locking), neo-Hooke
* frictional **contact** with the moving head (penalty), gravity, air drag
* implicit Euler as an incremental energy, **Newton with automatic differentiation**, static condensation, sparse Cholesky
* 69 426 unknowns, 800 time steps, 2–6 Newton iterations per step
* movie: NGSolve webgui, frames captured with headless Chrome

### What the note records from the iterations

* St. Venant–Kirchhoff lost convexity at the tips of the holes, Newton stalled: switched to neo-Hooke
* the heavy hem folded through itself, an element inverted: added air drag
* the mixed form needed a general LU: hybridization gives an SPD system for Cholesky
* one summed energy linearized 20$\times$ slower than separate terms

Full note: M. Neunteufel, J. Schöberl, *A Halloween ghost in NGSolve: nonlinear HHJ shell with frictional contact*, NGSolve Technical Note 2026-04.

## Use NGSolve with your own agent

NGSolve fits agents well:

* `pip install ngsolve`, runs headless, a model is a short Python script
* the code reads like the math: the agent writes `grad(u)*grad(v)*dx`, you can check it
* results are numbers the agent can test: compare with analytic solutions (harmonic oscillator, solenoid formula, Gauss–Bonnet), convergence under refinement
* pictures for you: the webgui renders in Jupyter, or headless to images and movies

How to work with it:

* start from a working example: the i-tutorials, the iFEM course, these notebooks
* give the agent the technical notes as context
* ask for a sanity check with every model, and for a convergence study before you trust a number
* run big jobs remotely: we use a skill that sends NGSolve scripts to the MUSICA cluster (H100 GPUs) via Slurm

*Talk to us:* you know the physics, we know the numerical methods and the library, and agents do much of the work in between.
