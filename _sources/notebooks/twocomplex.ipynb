# The 2-complex

The de Rham complex organizes the spaces for scalar and vector fields. Strains, stresses, metrics and curvature are matrix fields.
Their Sobolev spaces fit into a two-dimensional diagram, the 2-complex:

<img src="../pictures/twocomplex.png" width="80%">

with $\operatorname{def} u = \tfrac{1}{2} (\nabla u + \nabla u^T)$, $\operatorname{dev} \eta = \eta - \tfrac{1}{3} \operatorname{tr} \eta \, I$, and $\operatorname{curl}^T \gamma = (\operatorname{curl} \gamma^T)^T$.
The matrix fields are symmetric (${\mathbb S}$) or trace-free (${\mathbb T}$).

* first row and first column: the de Rham complex
* dual spaces (with flipped boundary conditions) are at opposite positions: $[ V^{k,l}]^\ast = V^{3-k, 3-l}$
* related to the BGG complexes of D. Arnold and K. Hu

J. Gopalakrishnan, K. Hu, J. Schöberl: *A 2-complex containing Sobolev spaces of matrix fields*, [arXiv:2507.11869](https://arxiv.org/abs/2507.11869)

## Applications

Every arrow is the differential operator of a physical model, and of a finite element method for it:

<img src="../pictures/twocomplex_applications.png" width="100%">

The diagonal arrows are second order operators: $\operatorname{hess} = \operatorname{def} \operatorname{grad}$, $\operatorname{inc} = \operatorname{sym} \operatorname{curl}^T \operatorname{curl}$, and $\operatorname{div} \operatorname{div}$.
They belong to the Hessian complex, the elasticity complex and the $\operatorname{div} \operatorname{div}$ complex, which are paths through the diagram.

## The n-dimensional 2-complex

<img src="../pictures/twocomplex_young.png" width="50%">

The spaces are represented by Young tableaux. They are spaces of double forms (Calabi 1961; Berchenko-Kogan and Gawlik 2025):

$$
\Lambda^k \otimes \Lambda^l = \{ f(\underbrace{X_1, \ldots , X_k}_{\text{skew}} ; \underbrace{Y_1, \ldots, Y_l}_{\text{skew}} ) \}
$$

The Bianchi operator $s : \Lambda^k \otimes \Lambda^l \rightarrow \Lambda^{k+1} \otimes \Lambda^{l-1}$ is

$$
\begin{aligned}
& (s f) (X_1, \ldots,  X_{k+1} ; Y_1, \ldots , Y_{l-1}) := \\
& \qquad \sum_{a=1}^{k+1} (-1)^{a-1} f(X_1, \ldots, \widehat{X_a}, \ldots, X_{k+1}; X_a, Y_1, \ldots,  Y_{l-1})
\end{aligned}
$$

* irreducible sub-spaces: $V^{k,l}  := \{ f \in  \Lambda^k \otimes \Lambda^l  : sf = 0 \}$
* differential operators: $d = d_1$ and $d_s = d_2 - s^* s d_1$
